Let's use the `marketing` dataset from `datarium` (Data Bank for Statistical Analysis and Visualization) package. It contains the impact of three advertising medias (youtube, facebook and newspaper) on sales. Data are the advertising budget in thousands of dollars along with the sales. The advertising experiment has been repeated 200 times with different budgets and the observed sales have been recorded. To illustrate linear regression with univariate (one variable), in this section we will only consider the impact of advertising budget of youtube on sales.

```r
# load libraries

library(datarium)
library(ggplot2)
library(broom) # creates a tidy data frame from statistical test results
install.packages("ggfortify")
library(ggfortify)
library(tidyverse)
```

```r
# Load and inspect the marketing data 
data("marketing", package = "datarium")
str(marketing)
head(marketing, 4)
```

```r
# Visualization 
ggplot(marketing, aes(x = youtube, y = sales)) +
  geom_point() #add points to a plot so get scatter plot here
```

```r
# Compute the correlation coefficient between the two variables
cor(marketing$sales, marketing$youtube)
```

It shows there is a strong correlation between `youtube` and `sales`. That is the preliminary to perform a simple linear regression analysis.

```r
model <- lm(sales ~ youtube, data = marketing)
model
```

The estimated regression line is $sales =8.43911 +0.04754 \times youtube$.

```r
model.arg <- augment(model)
head(model.arg)
```

* .fitted: fitted values (the value that the regression line gave for response variable)
* .resid: residual errors (distance between observed data point and regression line value)
* .hat: hat values, used to detect high-leverage points by calculating how much influence the observed data point has on its own fitted value. A value of .hat above $2(p + 1)/n$ indicates an observation with high leverage where p is number of predictor variables +1 for intercept (total # parameters you are estimating). 
* .std.resid: standardized residuals, which is the residuals divided by their standard errors. Used to detect outliers (or extreme values in the outcome y variable)
* .cooksd: Cook’s distance, used to detect influential values, which can be an outlier or a high leverage point

```r
ggplot(model.arg, aes(x = youtube, y = sales)) +
  geom_point()+
  stat_smooth(method = lm, se = FALSE) + #plots lm model
  geom_segment(aes(xend = youtube, yend = .fitted), color = "red", linewidth = 0.3) #plots
  #line segments between points where x and y above are starting points
```

The following plot allows us to check for the assumptions that need to be met for linear regression. 

##### Analysis of Residual Diagnostic Plots

1. **Residuals vs Fitted Plot**:
   - **Purpose**: This plot checks for linearity and homoscedasticity (constant variance).
   - **Ideal Situation**: The red line should be relatively horizontal, and the residuals should be randomly scattered around zero without forming clear patterns.
   - **Observation**: In the plot, the red line appears mostly horizontal but shows a slight curve. This suggests a potential minor issue with non-linearity. Additionally, if the residuals exhibit more spread at certain fitted values, that indicates heteroscedasticity (non-constant variance).

2. **Normal Q-Q Plot**:
   - **Purpose**: This plot assesses whether the residuals follow a normal distribution.
   - **Ideal Situation**: The points should fall approximately along the diagonal line.
   - **Observation**: There is some deviation from the diagonal line, particularly at the tails. This indicates that the residuals might not be perfectly normally distributed. Slight deviations are normal, but significant deviations may require transformation.

3. **Scale-Location Plot**:
   - **Purpose**: This plot checks for homoscedasticity, or constant variance of the residuals.
   - **Ideal Situation**: The red line should be flat, and the residuals should be evenly spread across the range of fitted values.
   - **Observation**: The red line appears relatively flat, but there are some variations towards the higher fitted values. This suggests that the variance might not be perfectly constant, indicating mild heteroscedasticity.

4. **Residuals vs Leverage Plot**:
   - **Purpose**: This plot identifies influential observations that may disproportionately affect the model fit.
   - **Ideal Situation**: Most points should lie within the dashed lines, indicating low leverage and influence.
   - **Observation**: Most of the points fall within the bounds of the Cook’s distance lines, and there are no clear indications of highly influential points. This suggests that no single observation is unduly influencing the model.
